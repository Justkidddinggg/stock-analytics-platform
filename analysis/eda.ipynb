```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from sqlalchemy import create_engine
import os
```

```python
BASE_DIR = os.path.dirname(os.path.dirname(os.path.abspath("__file__")))
DB_PATH = os.path.join(BASE_DIR, 'data', 'stock_data.db')
engine = create_engine(f'sqlite:///{DB_PATH}')
```

```python
df = pd.read_sql("SELECT * FROM stock_prices", con=engine)
df['date'] = pd.to_datetime(df['date'])
```

```python
print(df.shape)
display(df.head())
```

 Stock Price Trend

```python
fig = px.line(
    df, x='date', y='close', color='ticker',
    title='Stock Price Trend (2Y)',
    labels={'close': 'Price (USD)', 'date': 'Date'}
)
fig.show()
```

MA AAPL

```python
aapl = df[df['ticker'] == 'AAPL']
fig = go.Figure()
fig.add_trace(go.Scatter(x=aapl['date'], y=aapl['close'], name='Close', line=dict(color='gray', width=1)))
fig.add_trace(go.Scatter(x=aapl['date'], y=aapl['MA7'],   name='MA7',   line=dict(color='blue')))
fig.add_trace(go.Scatter(x=aapl['date'], y=aapl['MA30'],  name='MA30',  line=dict(color='orange')))

fig.update_layout(title='AAPL — Close Price vs Moving Averages')
fig.show()
```

Volatility Comparison

```python
fig = px.box(
    df, x='ticker', y='volatility', color='ticker',
    title='Volatility Comparison by Stock'
)
fig.show()
```

Correlation

```python
# Correlation ระหว่างหุ้นแต่ละตัว
pivot = df.pivot_table(index='date', columns='ticker', values='daily_return')
corr = pivot.corr()

fig = px.imshow(
    corr,
    text_auto='.2f',
    color_continuous_scale='RdBu_r',
    title='Correlation of Daily Returns between Stocks',
    zmin=-1, zmax=1
)
fig.show()
```
